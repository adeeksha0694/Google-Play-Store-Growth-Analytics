## Data Cleaning

```python
# imports

import pandas as pd

import numpy as np
```

### Reviews Column

### Size Col

```python
# Clean size column

df['Size'].head(20)
```

```python
df['Size'].value_counts().head(20)
```

```python
# checking size contains KB values

df[df['Size'].str.contains('k', case=False, na=False)]['Size'].head(20)
```

```python
# convert Size to MB

df['Size_MB'] = df['Size'].apply(
    lambda x: float(x[:-1])/1024 if x[-1].lower()=='k'
    else float(x[:-1]) if x[-1].lower()=='m'
    else None
)
```

```python
df[['Size', 'Size_MB']].head(20)
```

```python
df[df['Size'].str.contains('k', case=False, na=False)][['Size', 'Size_MB']].head(10)
```

```python
# replace size('Varies with device') into NaN

import numpy as np

df.loc[df['Size']=='Varies with device', 'Size_MB'] = np.nan

df['Size_MB'].isnull().sum()
```

```python
# Verify the Size_MB column

df[['Size', 'Size_MB']].tail(10)
```

### Installs Col

```python
# Inspecting

df['Installs'].head(20)
```

```python
# check unique values 

df['Installs'].unique()
```

```python
# Fix only the inconsistent 1000+

df['Installs']=df['Installs'].replace('1000+', '1,000+')
df['Installs'].unique()
```

```python
df['Installs'] = df['Installs'].str.replace(',', '', regex=False)

df['Installs'] = df['Installs'].str.replace('+', '', regex=False)

df['Installs'] = pd.to_numeric(df['Installs'])
```

```python
df['Installs'].dtype
```

```python
df['Installs'].head(20)
```

### Price Col

```python
# inspect price col

df['Price'].head(20)
```

```python
df['Price'].unique()
```

```python
# remove $ sign

df['Price'] = df['Price'].str.replace('$','',regex=False)
df['Price'].unique()
```

```python
# converting price to numeric

df['Price'] = pd.to_numeric(df['Price'])
df['Price'].dtype
```
